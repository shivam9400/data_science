```python
def walk(steps):
    for i in range(1, steps+1):
        print("let's take step {}".format(i))

walk(10)
```

```python
def recursive_walk(step):
    if step==0:
        return
    print("let's take step {}".format(step))
    recursive_walk(step-1)

recursive_walk(10)
```

```python
def recursive_walk_v1(step):
    if step==0:
        return
    recursive_walk_v1(step-1)
    print("let's take step {}".format(step))
    

recursive_walk_v1(10)
```
