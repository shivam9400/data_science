```python
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None
```

```python
class LinkedList:
    def __init__(self):
        self.head = None

    def __repr__(self):
        curr = self.head
        repr = ""
        while curr is not None:
            repr = repr + str(curr.value)
            if curr.next is not None:
                repr = repr + " --> "
            curr = curr.next
        return repr

    def __contains__(self, value):
        curr = self.head
        while curr.next is not None:
            if curr.value == value:
                return True
            curr=curr.next
        return False

    def __len__(self):
        len=0
        curr = self.head
        while curr is not None:
            len=len+1
            curr = curr.next
        return len

    def append(self, value):
        if self.head is None:
            self.head = Node(value)
        else:
            curr = self.head
            while curr.next:
                curr = curr.next
            curr.next = Node(value)

    def prepend(self, value):
        if self.head is None:
            self.head = Node(value)
        else:
            curr = Node(value)
            curr.next = self.head
            self.head = curr

    def insert(self, value, index):
        if index == 0:
            self.prepend(value)
        else:
            if self.head is None:
                raise ValueError("Index out of bounds")
            else:
                curr = self.head
                for i in range(index-1):
                    if curr.next is None:
                        raise ValueError("Index out of bounds")
                    curr=curr.next

                new_node = Node(value)
                new_node.next = curr.next   
                curr.next = new_node

    def delete(self, value):
        curr = self.head
        if curr is not None:
            if curr.value == value:
                self.head=curr.next
            else:
                while curr.next:
                    if curr.next.value == value:
                        curr.next=curr.next.next
                        break

    def reverse(self):
        curr = self.head
        prev_node = None

        while curr is not None:
            next_node = curr.next
            curr.next = prev_node
            prev_node=curr
            curr = next_node
            
        self.head = prev_node
```
