```python
# 2. Given a dictionary, return a list of its keys sorted by their values

data = {"a":5, "b":2, "c":3, "d":10}

keys = list(data.keys())
values = [data[k] for k in keys]
sorted_indices = np.argsort(values)

sorted_keys = [keys[index] for index in sorted_indices]

print(sorted_keys)
```

```python
# 3. Given an array of integers nums and an integer target, return indices of the two numbers 
# such that they add up to target.

# Input: nums = [2,7,11,15], target = 9
# Output: [0,1]

data = [2,7,11,15]
target = 9

store={}
for index, j in enumerate(data):
    diff = target-j
    if diff not in store:
        store[j]=diff
    else:
        print([index, data.index(diff)])
```

```python
# 4. Sort array using bubble sort algorithm

data = [3,4,1,8,7]

n = len(data)
for i in range(n):  # suppose i=3, then last 3 elements should already be sorted
    for j in range(0, n-i-1):      # j --> 0 to 1
        if data[j]>data[j+1]:
            data[j], data[j+1] = data[j+1], data[j]

print(data)
```

```python
# 5. Sort array using quick sort algorithm

data = [3,4,1,8,7]

def quick_sort(data):
    if len(data)<=1:
        return data
    pivot = data[0]   # pick first element as pivot
    left = [j for j in data[1:] if j<pivot]
    right = [j for j in data[1:] if j>=pivot]

    return quick_sort(left) + [pivot] + quick_sort(right)

quick_sort(data)
```

```python
# 6. Given an integer array nums, return all the triplets [nums[i], nums[j], nums[k]] 
# such that i != j, i != k, and j != k, and nums[i] + nums[j] + nums[k] == 0

original_nums = [-1,0,1,2,-1,-4]

nums = quick_sort(original_nums)
result = []

for index,elem in enumerate(nums):
    left = index+1
    right = len(nums)-1
    while left<right:
        print("Current values, \n\tindex --> {}, left_index --> {}, right_index --> {}".format(index, left, right))
        total = elem + nums[left] + nums[right]
        if total == 0:
            if [elem, nums[left], nums[right]] not in result:
                result.append([elem, nums[left], nums[right]])

            # check for any duplicates
            while (left<right) and (nums[left]==nums[left+1]):
                left = left+1
            while (left<right) and (nums[right]==nums[right-1]):
                right = right-1
            
            left = left+1
            right = right-1
        elif total < 0:
            left = left+1
        else:
            right = right-1


print(result)
```

```python
# 7. Given an integer array nums sorted in non-decreasing order, remove the duplicates in-place 
# such that each unique element appears only once. 
# The relative order of the elements should be kept the same. 

nums = [1,1,2]

store={}
i = 0
while i < len(nums):
    if nums[i] not in store:
        store[nums[i]]=1
        i=i+1   # move the pointer forward
    else:
        store[nums[i]]+=1
        nums.pop(i)  # no need to move pointer ahead

print(nums)
```

```python
# 10. Write function to reverse an first elements of an array, given start and end index

nums = [1,2,3,4,5,6,7]

def reverse_array(nums, left, right):
    while left<right:
        nums[left], nums[right] = nums[right], nums[left]
        left=left+1
        right=right-1
    
    return nums

reverse_array(nums, 0, len(nums)-1)
```

```python
# 11. Use reverse_array to rotate array to the right by k steps, where k is non-negative.

nums = [1,2,3,4,5,6,7]
k = 3

if k>len(nums):
    k=k%len(nums)
    
nums =  reverse_array(nums, 0, len(nums)-1)
print(nums)

# reverse first k elements
nums =  reverse_array(nums, 0, k-1)
print(nums)

# reverse remaining elements
nums =  reverse_array(nums, k, len(nums)-1)
print(nums)
```

```python
# 12. Given an array of positive integers nums and a positive integer target, 
# return the minimal length of a subarray whose sum is greater than or equal to target. 

nums = [2,3,1,2,4,3]
target = 7

n = len(nums)
left = 0
total = 0 #current total sum of the window
min_len = float("inf")

for right in range(n):
    total = total + nums[right]

    while total>=target:
        if right-left+1<min_len:
            start, end = left, right
        min_len = min(min_len, right-left+1)
        total = total - nums[left]
        left=left+1

print(nums[start:end+1])
print(min_len)
```

```python
# 13. Given an array of positive integers nums and a positive integer target, 
# return the maximum length of a subarray whose sum is greater than or equal to target. 

nums = [2,3,2,2,4,3]
target = 7

n = len(nums)
left=0
total = 0  # current value of the sum
max_len = 0

for right in range(n):
    total = total + nums[right] # as we iterate in the for loop, window keeps spanning and 
                                # total value keeps increasing
    
    # but, we have to control when total goes greater than target
    while total >= target:
        if right-left+1 > max_len:
            start, end =  left, right
            max_len = right-left+1
        total = total - nums[left]
        left=left+1
    
print(nums[start:end+1])
```

```python
# 14. You are given an array of integers nums, there is a sliding window of size k which is moving 
# from the very left of the array to the very right. You can only see the k numbers in the window. 
# Each time the sliding window moves right by one position. Return the max sliding window.

nums = [1,3,-1,-3,5,3,6,7]
k = 3

left = 0
right=k-1
n = len(nums)
result = []

print("length : {}".format(n))

while right<n:
    print("window --> {}".format(nums[left:right+1]))
    print("\t right --> {}".format(right))
    result.append(int(np.max(nums[left:right+1])))
    left=left+1
    right=right+1
    
print(result)
```
