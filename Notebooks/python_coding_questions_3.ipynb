```python
import pandas as pd
import numpy as np
```

```python
# 1. who is the top-performing salesperson per region.

data = {
    'Region': ['North', 'South', 'East', 'West', 'North', 'South', 'East', 'West'],
    'Salesperson': ['Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Frank', 'Grace', 'Heidi'],
    'Product': ['A', 'A', 'B', 'B', 'A', 'B', 'A', 'B'],
    'Revenue': [1200, 1500, 800, 950, 1100, 1300, 1000, 1050]
}
df = pd.DataFrame(data)

# step 1
grouped = df.groupby(['Region', 'Salesperson'])['Revenue'].sum().reset_index()
print(grouped)

# step 2
grouped.sort_values(['Region', 'Revenue'], ascending=[True, False]).groupby(['Region']).first()
```

```python
# 2. what is the top performing product per region

data = {
    'Region': ['North', 'South', 'East', 'West', 'North', 'South', 'East', 'West'],
    'Salesperson': ['Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Frank', 'Grace', 'Heidi'],
    'Product': ['A', 'A', 'B', 'B', 'A', 'B', 'A', 'B'],
    'Revenue': [1200, 1500, 800, 950, 1100, 1300, 1000, 1050]
}
df = pd.DataFrame(data)
display(df)

# step 1
grouped = df.groupby(['Region', 'Product'])['Revenue'].sum().reset_index()
display(grouped)

# step 2
grouped.sort_values(['Region', 'Revenue'], ascending=[True, False]).groupby('Region').first()
```

```python
# 3. compute multiple aggregations like mean, sum, and count per group using pandas.groupby
#### Compute the mean, sum, and count of Revenue per Region.

# Sample sales data
data = {
    'Region': ['North', 'South', 'North', 'East', 'South', 'East', 'North'],
    'Salesperson': ['Alice', 'Bob', 'Alice', 'Charlie', 'Bob', 'Charlie', 'Eve'],
    'Revenue': [200, 150, 300, 400, 250, 100, 500]
}

df = pd.DataFrame(data)

df.groupby('Region')['Revenue'].agg(['mean','sum', 'count'])
```

```python
# 4. join the dataframes

# Customer information
customers = pd.DataFrame({
    'CustomerID': [101, 102, 103, 104],
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'City': ['Pune', 'Delhi', 'Mumbai', 'Chennai']
    }
    )

display(customers)
# Order information
orders = pd.DataFrame({
    'OrderID': [1001, 1002, 1003, 1004, 1005],
    'CustomerID': [101, 102, 101, 105, 103],
    'Amount': [250, 450, 300, 500, 150]
})

display(orders)
display(pd.merge(customers, orders, 
         on = 'CustomerID', how='outer'))

display(pd.merge(customers, orders, 
         on = 'CustomerID', how='inner'))
```
