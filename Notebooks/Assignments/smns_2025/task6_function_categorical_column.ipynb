# Task 6: Functions for Categorical attributes

```python
from collections import Counter

def group_categories(categories, threshold_percentage, generic_value='ABC'):
    # if empty list, return empty list
    if not categories:
        return []

    # if threshold is out of bounds, raise error
    if not (0 <= threshold_percentage <= 100):
        raise ValueError("Threshold percentage must be between 0 and 100.")

    # Calculate the frequency of each category
    category_counts = Counter(categories)
    total_count = len(categories)

    # Convert threshold to a proportion
    threshold_proportion = threshold_percentage / 100.0

    # Determine which categories are infrequent
    infrequent_categories = set()
    for category, count in category_counts.items():
        frequency = count / total_count
        if frequency < threshold_proportion:
            infrequent_categories.add(category)

    # Create the output list, replacing infrequent categories
    modified_categories = []
    for category in categories:
        if category in infrequent_categories:
            modified_categories.append(generic_value)
        else:
            modified_categories.append(category)

    return modified_categories
```

```python
data = ['USA', 'CAN', 'USA', 'MEX', 'USA', 'CAN', 'MEX', 'CHN', 'FRA', 'USA']
threshold = 25.0

result = group_categories(data, threshold)
print(f"Modified Data: {result}")
```

## Utility `group_categories()` function in constructing a logistic regression model

> - mitigating issues related to high cardinality (lot of unique value in the categorical column. However, only few of them appear frequently. Rest are rare appearances.) in categorical features.
>   - When we convert such column to numerical column for logistic regression using one-hot encoding, each unique category value becomes its own binary feature (column). If a value only appears a few times, its corresponding column will be mostly zeros, leading to a **sparse matrix**.
> - By leveraging `group_categories()` function, we can significantly reduce the number of resulting features after one-hot encoding. This makes the dataset denser and the model simpler.
>   - It also helps in better generalization by the model.
