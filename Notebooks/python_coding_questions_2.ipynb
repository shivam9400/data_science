# String based

## Reverse a string

```python
### using for loop

data = "silicon"

n = len(data)
reversed_str = ""   # initialize variable to store the reversed string
for i in range(n):
    reversed_str += data[n - 1 - i]

print(reversed_str)
```

```python
### using while loop

data = "silicon"

index = len(data)-1
reverse_str = ""

while index>=0:
    reverse_str = reverse_str + data[index]
    index = index-1

print(reverse_str)
```

```python
### using recursion

data = "silicon"

def reverse(text):
    if len(text) <=1:
        return text
    return reverse(text[1:])+text[0]  # --> ilicon + s ---> licon + i + s ---> icon + l + i + s

print(reverse(data))
```

## Longest Substring Without Repeating Characters

```python
### using two pointer - sliding window approach

data = "aabcdda"


start = 0
max_len = 0
max_substring = ""
seen = set()  # To store characters in the current window

for end in range(len(data)):
    print("End pointer : {}".format(end))
    
    # If we see a duplicate character
    while data[end] in seen:
        print("\t start: {}".format(start))
        seen.remove(data[start])  # Remove the leftmost character
        start += 1             # Move the window forward

    seen.add(data[end])  # Add the current character to the set
    print("\t seen: {}".format(seen))
    
    # Update max length and substring if needed
    if end - start + 1 > max_len:
        max_len = end - start + 1
        max_substring = data[start:end+1]
        print("\t max_substring: {}".format(max_substring))
```

## Compress a string ("aaabba" → "a3b2a")

```python
data = "aaabba"
n = len(data)
index =0
count = 0

comp_str=""

while index<n:
    
    # handling first element
    if index==0:
        comp_str = comp_str + data[index]
        index+=1
        count = count+1
        continue

    # handling other elements
    else:
        if data[index]==comp_str[-1]:    # if incoming is "b", already string is "ab"
            count=count+1
        else:
            comp_str = comp_str + str(count)
            comp_str = comp_str + data[index]
            count=1
        index = index+1

print(comp_str)
```

## find maximum element in an array

## Reverse a list using two pointers

```python
data = [3,4,5,6,2]

n = len(data)

left = 0
right = n-1

while left<right:
    swap = data[left]
    data[left] = data[right]
    data[right] = swap

    left = left+1
    right = right-1
    
print(data)
```

## Sort a list using two for loops

```python
data = [3,2,1,4,5]

n = len(data)

for i in range(n):
    for j in range(0, n-i-1): # In each pass (outer loop i), the largest unsorted element "bubbles up" to its correct position at the end of the list.
                              # So after the first pass, the last element is already sorted.
        if data[j] > data[j+1]:
            data[j], data[j+1] = data[j+1], data[j]

print(data)
```

## Cyclically rotate a list by one using two pointers

```python
data = [3, 4, 2, 5, 6]

left = 0
n = len(data)

right = n-1

while left<right:    
    swap = data[left]
    data[left] = data[right]
    data[right] = swap

    left =  left+1

print(data)
```

## Merge two sorted arrays using 2 pointers

> - Initialize two pointers: i = 0 for arr1, and j = 0 for arr2.
> - Compare elements at both pointers.
> - Append the smaller one to the result and move that pointer forward.
> - Once one array is exhausted, append the remaining elements from the other.

```python
data1 = [1,3,4]
data2 = [2,5,6]
sorted_list = []

left1 = 0
left2 = 0

while left1 < len(data1) and left2 < len(data2):
    if data1[left1] < data2[left2]:
        sorted_list.append(data1[left1])
        left1 = left1 + 1
    else:
        sorted_list.append(data2[left2])
        left2 = left2 + 1

while left1 < len(data1):
    sorted_list.append(data1[left1])
    left1 = left1 + 1

while left2 < len(data2):
    sorted_list.append(data2[left2])
    left2 = left2 + 1
    
print(sorted_list)
```

## Find the intersection of two sorted lists using two pointers

```python
data1 = [1,3,4]
data2 = [2,3,6]

intersection_list = []

left1 = 0
left2 = 0

while left1<len(data1) and left2< len(data2):
    if data1[left1]==data2[left2]:
        intersection_list.append(data1[left1])
        left1 = left1+1
        left2 = left2+1
        
    elif data1[left1]<data2[left2]:  # data1[left1] is smaller, so it can't be in the intersection (since data2[left2] is bigger and sorted)
        left1 =left1+1
        
    else:
        left2=left2+1
        
print(intersection_list)
```

## 2 Sum Problem

> Given a list of n integers and a target value, check if there exists a pair whose sum equals the target.

```python
data = [2, 4,6,3,2,1]
target = 6

store = {}

for j in data:
    diff = target-j
    
    if diff not in store:   # why search for comp? --> because if composite number came during iteration, diff will already be present as key in store 
        store[j] = diff
    else:
        print(j, diff)
```
