# Importing Libraries

```python
from serpapi import GoogleSearch
import os
import dspy
```

# Checking if LM is working

```python
lm = dspy.LM('ollama_chat/llama3.2', 
             api_base='https://preeminent-untactually-tora.ngrok-free.dev', 
             api_key='')
# 2. Set it as the global default
dspy.configure(lm=lm)

# 3. Use DSPy as usual
qa = dspy.Predict("question -> answer")
response = qa(question="What is the capital of France?")
print(response.answer)
```

# Checking if SERP API is working

```python
params = {
    "engine": "google_images",
    "q": "Men's Nike Air Max 270 Black and White",
    "api_key":os.getenv("SERP_API_KEY"),
    "location": "United States",
    "hl": "en",
    "gl": "us"
}

search = GoogleSearch(params)
results = search.get_dict()
results
```

```python
# Access the first few image results
if "images_results" in results:
    for image in results["images_results"][:5]:
        print(f"Title: {image['title']}")
        print(f"Image URL: {image['original']}")
        print(f"Thumbnail: {image['thumbnail']}\n")
```

```python
from IPython.display import Image, display, HTML
image_urls = []

if "images_results" in results:
    for image in results["images_results"][:5]:
        image_urls.append(image['original'])
```

```python
# Display them in a nice row
images_html = "".join([f'<img src="{url}" style="display:inline-block; margin:5px; width:200px;"/>' for url in image_urls])
display(HTML(images_html))
```
