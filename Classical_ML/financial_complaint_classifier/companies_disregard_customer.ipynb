```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_pickle('cleaned_data.pkl')
```

```python
df['Lag_days'] = (df['Date received'] - df['Date sent to company']).dt.days
```

```python
company_stats = df.groupby('Company').agg({
    'Complaint ID': 'count',
    'Consumer disputed?': 'mean',
    'Timely response?': 'mean',
    'Lag_days': ['mean', 'median', 'max']
})

print(company_stats)
```
